## Q6. Median of horsepower — **Yes, it decreased**

Mediana original = 254.0; moda = 252.0; mediana após `fillna(moda)` = 252.0.

```python
med_antes = df['horsepower'].median()
moda = df['horsepower'].mode()[0]
med_depois = df['horsepower'].fillna(moda).median()
med_antes, moda, med_depois
```

## Q7. Sum of weights — **0.369** (0.36919696904925486)

```python
X = asia[['vehicle_weight', 'model_year']].head(7).to_numpy()
XTX = X.T @ X
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = np.linalg.inv(XTX) @ X.T @ y
w.sum()
```
